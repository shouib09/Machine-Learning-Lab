```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Sample time-series data
days = np.array([1,2,3,4,5,6,7]).reshape(-1,1)
sales = np.array([200,220,250,270,300,320,350])

model = LinearRegression()
model.fit(days,sales)

pred = model.predict(days)

print("MAE:",mean_absolute_error(sales,pred))
print("RMSE:",np.sqrt(mean_squared_error(sales,pred)))

plt.plot(days,sales,label="Actual")
plt.plot(days,pred,label="Predicted")
plt.legend()
plt.show()

# Residual plot
plt.scatter(days,sales-pred)
plt.title("Residual Errors")
plt.show()
```
